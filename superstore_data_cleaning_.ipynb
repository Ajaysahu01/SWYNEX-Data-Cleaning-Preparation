```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/superstore data.csv')
df
```

```python
df.size
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18'])
df
```

```python
df
```

```python
# Replace placeholder strings with actual NaN values, then drop them
df = df.replace(['ERROR', 'error', 'unknown', 'N/A'], np.nan)
df = df.dropna()
df
```

```python
df.shape
```

```python
df.columns = df.columns.str.strip()
```

```python
df
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
print("Rows and Columns:", df.shape)
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
df.select_dtypes(include='object').head()

```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed').dt.strftime('%d-%m-%y')
df.head()
```

```python
df[['Order_Date']]
```

```python
df.describe(include="all")
```

```python
numeric_cols = ['Quantity', 'Unit_Price', 'Discount', 'Sales_Amount', 'Cost_Amount']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Now describe the numeric columns
df.describe(include='number')
```

```python
df.info()
```

```python
df.to_excel('Superstore_Cleaned.xlsx', index=False)
```
